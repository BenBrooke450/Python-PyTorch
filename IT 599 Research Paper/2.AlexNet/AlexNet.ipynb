```python
import torch
import torch.nn as nn

class AlexNet(nn.Module):
    def __init__(self, num_classes):
        super(AlexNet, self).__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 64, kernel_size=3, stride=1, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size=2, stride=2),


            nn.Conv2d(64, 192, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size=2, stride=2),


            nn.Conv2d(192, 384, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),


            nn.Conv2d(384, 256, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),


            nn.Conv2d(256, 256, kernel_size=3, padding=1),
            nn.ReLU(inplace=True),
            nn.MaxPool2d(kernel_size=2, stride=2),
        )

        self.classifier = nn.Sequential(
            nn.Dropout(p=0.5),
            nn.Linear(256 * 4 * 4, 4096),
            nn.ReLU(inplace=True),
            nn.Dropout(p=0.5),
            nn.Linear(4096, 4096),
            nn.ReLU(inplace=True),
            nn.Linear(4096, num_classes),
        )

    def forward(self, x):
        x = self.features(x)
        x = torch.flatten(x, 1)
        x = self.classifier(x)
        return x


model = AlexNet(num_classes=10)
print(model)
```

```python
#IT 599 Research Paper/1.LeNet-5/data/cifar-10-batches-py
```

```python
loss_fn = nn.CrossEntropyLoss()
```

```python
optimiser = torch.optim.Adam(model.parameters(),lr=0.0001)
```

```python
from torchvision import transforms
import torchvision

transform = transforms.Compose([
    transforms.Resize((32, 32)),
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))])

"""transform_train = transforms.Compose([
    transforms.RandomHorizontalFlip(p=0.5), # 50% chance to flip
    transforms.RandomRotation(15),          # Rotate by up to 15 degrees
    transforms.RandomAffine(degrees=0, translate=(0.1, 0.1)), # Shift image slightly
    transforms.ToTensor(),
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
])"""
```

```python
shared_root='/Users/benjaminbrooke/PycharmProjects/Python_PyTroch/IT 599 Research Paper/1.LeNet-5/data/'
```

```python
train_set = torchvision.datasets.CIFAR10(root=shared_root, train=True, download=True, transform=transform)
```

```python
print(next(iter(train_set)))
```

```python
from torch.utils.data import DataLoader

train_loader = DataLoader(train_set,shuffle=True,batch_size=50)
```

```python
from ptflops import get_model_complexity_info
macs, params = get_model_complexity_info(model, (3, 32, 32), as_strings=True, print_per_layer_stat=False)
print(f"Computational complexity: {macs}")
print(params)
```

```python
from tqdm.notebook import tqdm
import time
from torch.utils.tensorboard import SummaryWriter
LeNet5_Metrics = SummaryWriter()

latency_per_image = []
global_i = 0

pbar = tqdm(train_loader)

all_preds = []
all_labels = []

for image,label in pbar:

    start_time = time.time()

    optimiser.zero_grad()

    y_pred =  model(image)

    loss = loss_fn(y_pred,label)

    loss.backward()

    optimiser.step()

    _, predicted = torch.max(y_pred.data, 1)
    correct = (predicted == label).sum().item()
    accuracy = correct / label.size(0)

    end_time = time.time()

    latency_per_image.append(end_time - start_time)

    LeNet5_Metrics.add_scalar("Loss/train - AlexNet IT599 Ben", loss.item(), global_i)
    LeNet5_Metrics.add_scalar("Accuracy/train - AlexNet IT599 Ben", accuracy, global_i)

    all_preds.extend(predicted.numpy())
    all_labels.extend(label.numpy())

    global_i += 1


LeNet5_Metrics.close()

avg_latency_batch = sum(latency_per_image) / len(latency_per_image)
avg_latency_image = avg_latency_batch / train_loader.batch_size
throughput = 1 / avg_latency_image

print(f"Latency: {avg_latency_image*1000:.2f} ms | Throughput: {throughput:.2f} items/sec")
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(all_labels, all_preds))
```
